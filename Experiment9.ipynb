```python
import matplotlib.pyplot as plt

# Sample dataset

months = ["Jan", "Feb", "Mar", "Apr", "May"]

production = [250, 280, 300, 270, 320]

workers = [20, 22, 25, 21, 27]

# 1. Line Plot

plt.figure(figsize=(8, 5))

plt.plot(months, production, marker="o")

plt.xlabel("Month")

plt.ylabel("Production")

plt.title("Production Trend - Line Plot")

plt.grid(False)

plt.show()


# 2. Bar Chart

plt.figure(figsize=(8, 5))

plt.bar(months, production)

plt.xlabel("Month")

plt.ylabel("Production")

plt.title("Monthly Production - Bar Chart")

plt.grid(False)

plt.show()


# 3. Scatter Plot

plt.figure(figsize=(8, 5))

plt.scatter(workers, production)

plt.xlabel("Number of Workers")

plt.ylabel("Production")

plt.title("Production vs Workers - Scatter Plot")

plt.grid(False)

plt.show()


# 4. Histogram

plt.figure(figsize=(8, 5))

plt.hist(production, bins=5, edgecolor="black")

plt.xlabel("Production")

plt.ylabel("Frequency")

plt.title("Production Distribution - Histogram")

plt.grid(False)

plt.show()


# 5. Pie Chart

plt.figure(figsize=(8, 5))

plt.pie(
    production,
    labels=months,
    autopct="%1.1f%%"
)

plt.title("Production Distribution - Pie Chart")

plt.show()
```
