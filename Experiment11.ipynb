```python
import pandas as pd

# Create student dataset

data = {

    "Student_ID": [101, 102, 103, 104, 105],

    "Name": ["Arun", "Bala", "Charan", "Divya", "Kavin"],

    "Marks": [78, 85, 72, 91, 88]

}

df = pd.DataFrame(data)

# Display dataset

print("Student Data:")

print(df)

# Display first rows

print("\nFirst Rows:")

print(df.head())

# Display information

print("\nData Information:")

df.info()

# Display marks column

print("\nMarks:")

print(df["Marks"])

# Basic statistics

print("\nSummary Statistics:")

print(df.describe())

# Marks statistics

print("\nAverage Marks:", df["Marks"].mean())

print("Median Marks:", df["Marks"].median())

print("Maximum Marks:", df["Marks"].max())

print("Minimum Marks:", df["Marks"].min())
```
