```python
import numpy as np

import pandas as pd

from sklearn.preprocessing import StandardScaler, MinMaxScaler

# New dataset

data = {

    "Age": [18, 20, 22, 24, 26],

    "Income": [15000, 20000, 25000, 30000, 35000],

    "Experience": [1, 2, 3, 5, 7]

}

df = pd.DataFrame(data)

print("Original Dataset:")

print(df)

# Standard Scaling

sc = StandardScaler()

standard_scaled = sc.fit_transform(df)

print("\nStandard Scaled Data:")

print(standard_scaled)

# Min-Max Scaling

mms = MinMaxScaler()

minmax_scaled = mms.fit_transform(df)

print("\nMin-Max Scaled Data:")

print(minmax_scaled)
```
