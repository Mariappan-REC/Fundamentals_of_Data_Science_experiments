```python
import pandas as pd
import numpy as np

# New dataset

data = {
    "Department": ["CSE", "ECE", "IT", "CSE", "EEE",
                   "IT", "ECE", "EEE", np.nan, "CSE"],

    "Age": [21, 22, 20, 23, 24, np.nan, 22, 25, 21, 23],

    "Salary": [30000, 35000, 32000, np.nan, 40000,
               33000, 37000, 45000, 34000, 38000],

    "Selected": ["Yes", "No", "Yes", "Yes", "No",
                 "No", "Yes", "No", "Yes", "No"]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# Handle missing values

df["Department"] = df["Department"].fillna(
    df["Department"].mode()[0]
)

df["Age"] = df["Age"].fillna(df["Age"].median())

df["Salary"] = df["Salary"].fillna(
    round(df["Salary"].mean())
)

# Convert Department into numerical columns

department = pd.get_dummies(df["Department"])

# Create preprocessed dataset

updated_dataset = pd.concat(
    [department, df[["Age", "Salary", "Selected"]]],
    axis=1
)

# Convert Selected into 0 and 1

updated_dataset["Selected"] = updated_dataset["Selected"].replace(
    {"No": 0, "Yes": 1}
)

print("\nPreprocessed Dataset:")
print(updated_dataset)

print("\nMissing Values:")
print(updated_dataset.isnull().sum())
```
