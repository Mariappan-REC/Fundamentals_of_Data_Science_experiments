```python
import pandas as pd
import matplotlib.pyplot as plt

# Create mobile app download data
data = {
    "Date": [
        "2026-03-01",
        "2026-03-02",
        "2026-03-03",
        "2026-03-04",
        "2026-03-05",
        "2026-03-06",
        "2026-03-07"
    ],
    "Downloads": [120, 150, 135, 180, 165, 210, 195]
}

# Create DataFrame
df = pd.DataFrame(data)

# Convert Date column to datetime
df["Date"] = pd.to_datetime(df["Date"])

# Set Date as index
df.set_index("Date", inplace=True)

# Display the data
print("Mobile App Download Data:")
print(df)

# Display basic information
print("\nStatistical Summary:")
print(df.describe())

# Calculate average downloads
print("\nAverage App Downloads:", df["Downloads"].mean())

# Find maximum downloads
print("Maximum App Downloads:", df["Downloads"].max())

# Find minimum downloads
print("Minimum App Downloads:", df["Downloads"].min())

# Plot the time series
plt.figure(figsize=(10, 5))

plt.plot(
    df.index,
    df["Downloads"],
    marker="o"
)

plt.title("Time Series Analysis of App Downloads")
plt.xlabel("Date")
plt.ylabel("Number of Downloads")
plt.xticks(rotation=45)
plt.grid(False)
plt.tight_layout()

plt.show()
```
