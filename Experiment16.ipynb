```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data = {
    "study_hours": [2, 3, 5, 6, 7, 4, 8, 9, 5, 10],
    "marks": [40, 45, 60, 65, 75, 55, 80, 90, 62, 95],
    "gender": ["Female", "Male", "Male", "Female", "Female",
               "Male", "Female", "Male", "Female", "Male"],
    "sports": ["No", "Yes", "No", "Yes", "No",
               "Yes", "No", "Yes", "No", "Yes"],
    "department": ["CSE", "ECE", "CSE", "IT", "ECE",
                   "IT", "CSE", "ECE", "IT", "CSE"],
    "result": ["Pass", "Pass", "Pass", "Pass", "Pass",
               "Pass", "Pass", "Pass", "Pass", "Pass"],
    "attempts": [2, 3, 2, 1, 2, 3, 1, 1, 2, 1]
}

students = pd.DataFrame(data)

print("Dataset:")
print(students)

# Quantitative Analysis

print("\nQuantitative Analysis:")
print(students.describe())

# Qualitative Analysis

print("\nQualitative Analysis:")
print(students["result"].value_counts())
print(students["department"].value_counts())

# Distribution of Marks

plt.hist(students["marks"], bins=5, edgecolor="black")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.title("Distribution of Marks")
plt.show()

# Marks vs Study Hours

plt.scatter(students["study_hours"], students["marks"])
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Marks vs Study Hours")
plt.show()

# Marks vs Study Hours with trend line

plt.scatter(students["study_hours"], students["marks"])

m, b = np.polyfit(students["study_hours"], students["marks"], 1)

plt.plot(students["study_hours"], m * students["study_hours"] + b)

plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Marks vs Study Hours with Regression Line")
plt.show()

# Box plot

plt.boxplot(students["marks"])
plt.ylabel("Marks")
plt.title("Box Plot of Marks")
plt.show()

# Bar chart for Result

students["result"].value_counts().plot(kind="bar")
plt.xlabel("Result")
plt.ylabel("Count")
plt.title("Students by Result")
plt.show()
```
