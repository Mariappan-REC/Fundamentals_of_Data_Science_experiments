```python
import matplotlib.pyplot as plt

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']

Product_A = [120, 150, 180, 160, 210, 250]
Product_B = [100, 140, 160, 190, 200, 230]

plt.plot(months, Product_A, marker='o', label='Product A')
plt.plot(months, Product_B, marker='o', label='Product B')

plt.title("Monthly Sales Comparison")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()

plt.show()
```

```python
import matplotlib.pyplot as plt

subjects = ['Maths', 'Physics', 'Chemistry', 'English', 'Computer']
marks = [85, 78, 92, 88, 95]

plt.bar(subjects, marks)

plt.title("Subject Marks")
plt.xlabel("Subjects")
plt.ylabel("Marks")

plt.show()
```
